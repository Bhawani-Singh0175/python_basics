```python
foods=[]
price=[]
total=0
while True:
    food=input("enter your food {q to quit}:")
    if food.lower() =="q":
        break
    else:
        prices=float(input(f"enter the price of {foods}:$"))
        foods.append(food)
        price.append(prices)
print("-----YOUR CART-----")
for f in foods:
    print(f)
for p in price:
    total=total+p
print(f"Total Amount is ${total}")
        
        
```
